```python
import random

# --- Input Section ---
# You can change N and the list below to test any custom state!
N = 4
custom_initial_state = [1,1,0,2] # Example state: Q in row 1 col 0, row 3 col 1, etc.
# ----------------------

def conflicts(state):
    count = 0
    for i in range(len(state)):
        for j in range(i + 1, len(state)):
            if state[i] == state[j]:
                count += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                count += 1
    return count

def get_neighbors(state):
    neighbors = []
    size = len(state)
    for col in range(size):
        for row in range(size):
            if row != state[col]:
                new_state = state.copy()
                new_state[col] = row
                neighbors.append(new_state)
    return neighbors

def hill_climbing(initial_state=None):
    if initial_state is None:
        state = [random.randint(0, N-1) for _ in range(N)]
    else:
        state = initial_state.copy()

    initial_conflicts = conflicts(state)
    print("Initial State:", state, f"(Conflicts: {initial_conflicts})")

    step = 0
    while True:
        current = conflicts(state)
        if current == 0:
            return state

        neighbors = get_neighbors(state)
        best = min(neighbors, key=conflicts)
        best_conflicts = conflicts(best)

        step += 1
        print(f"Step {step}: Best neighbor has {best_conflicts} conflicts")

        if best_conflicts >= current:
            print(f"Stuck in local optimum (Current conflicts: {current}, Best neighbor conflicts: {best_conflicts})")
            return None
        state = best

# Run with your custom input
solution = hill_climbing(initial_state=custom_initial_state)

if solution:
    print("\nSolution:", solution, "(Conflicts: 0)")
    for row in range(len(solution)):
        for col in range(len(solution)):
            if solution[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
else:
    print("\nNo solution found")
```
