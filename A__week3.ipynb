```python
import heapq

goal="123804765"

def h(s):
    count=0
    for i in range(9):
        if s[i]!='0' and s[i]!=goal[i]:
            count+=1
    return count

def print_matrix(state):
    for i in range(0, 9, 3):
        print(" ".join(state[i:i+3]))

def astar(start):
    pq = [(h(start),0,start)]
    visited=set()

    while pq:
        f,g,state=heapq.heappop(pq)

        if state in visited:
            continue
        visited.add(state)

        print(f"g={g},h={h(state)},f ={f}")
        print_matrix(state)
        print("-" * 9)

        if state == goal:
            print("Goal reached!")
            return

        zero=state.index('0')

        moves=[-3, 3, -1, 1]

        for move in moves:
            new = zero + move

            if 0<=new<9:

                if move==-1 and zero%3==0:
                    continue
                if move==1 and zero%3==2:
                    continue

                s=list(state)
                s[zero],s[new]=s[new],s[zero]
                s=''.join(s)

                if s not in visited:
                    new_g=g + 1
                    new_f=new_g + h(s)
                    heapq.heappush(pq,(new_f, new_g, s))


start = "283164705"
astar(start)
```
